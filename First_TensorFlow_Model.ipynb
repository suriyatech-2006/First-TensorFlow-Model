Train a small TensorFlow model to learn the relationship **y = 2x** and compare predicted and actual values.

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)
```

## 1. Create Training Data

The target relationship is **y = 2x**.

```python
x = np.array([1, 2, 3, 4, 5], dtype=np.float32)
y = np.array([2, 4, 6, 8, 10], dtype=np.float32)

print("x:", x)
print("y:", y)
```

## 2. Build the TensorFlow Model

A single Dense neuron is enough to learn the linear relationship.

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.05),
    loss='mse'
)

model.summary()
```

## 3. Train the Model

```python
history = model.fit(x, y, epochs=200, verbose=0)
print("Training completed.")
print("Final loss:", history.history['loss'][-1])
```

## 4. Compare Actual and Predicted Values

```python
predicted = model.predict(x, verbose=0).flatten()

print("Actual vs Predicted")
print("-------------------")
for actual, prediction in zip(y, predicted):
    print(f"Actual: {actual:.2f}   Predicted: {prediction:.2f}")
```

## 5. Test a New Value

The model should learn that when x = 6, y should be approximately 12.

```python
new_x = np.array([[6.0]], dtype=np.float32)
new_prediction = model.predict(new_x, verbose=0)[0][0]

print(f"For x = 6, predicted y = {new_prediction:.2f}")
print("Expected y = 12.00")
```

## 6. Plot Actual vs Predicted Values

```python
plt.figure(figsize=(7, 5))
plt.plot(x, y, 'o-', label='Actual y = 2x')
plt.plot(x, predicted, 's--', label='Predicted')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Actual vs Predicted Values')
plt.legend()
plt.grid(True)
plt.show()
```

## Result

The TensorFlow model learns the linear relationship **y = 2x**. The predicted values should be very close to the actual values, and the prediction for x = 6 should be close to 12.
